**Run order:** top to bottom. Pick the pair in the config-cell dropdowns —
the verdict, neighbour and contribution cells update automatically; re-run the
chunk cells after changing the selection. If the widgets can't render in your
viewer, edit `FILE_A` / `FILE_B` in the config cell instead.

```python
import json
import sys
from pathlib import Path

import numpy as np

sys.path.insert(0, str(Path.cwd()))
import embed_indicators as ei  # reuse strip_comments / chunk_source / center / embed fns

DBS = {
    '0.6b': Path('indicator_embeddings.json'),
    '4b': Path('indicator_embeddings_qwen3-embedding_4b.json'),
}

MODELS = {}       # label -> {script name: centered unit vector}
MEANS = {}        # label -> frozen reference mean
MODEL_NAMES = {}  # label -> ollama model name
for label, path in DBS.items():
    if not path.exists():
        print(f'{label}: {path.name} not found — skipped')
        continue
    db = json.loads(path.read_text())
    if db.get('preprocess') != ei.PREPROCESS_VERSION or 'reference_mean' not in db:
        print(f'{label}: {path.name} is stale (preprocess={db.get("preprocess")}, '
              f'need {ei.PREPROCESS_VERSION}) — skipped. Rebuild with:\n'
              f'    .venv/bin/python embed_indicators.py --model {db["model"]} build')
        continue
    mu = np.asarray(db['reference_mean'])
    MEANS[label], MODEL_NAMES[label] = mu, db['model']
    MODELS[label] = {n: ei.center(np.asarray(e['vector']), mu) for n, e in db['indicators'].items()}
    print(f'{label}: loaded {len(MODELS[label])} vectors from {path.name} '
          f'(model={db["model"]}, |reference_mean|={np.linalg.norm(mu):.3f})')

all_names = sorted(set().union(*(v.keys() for v in MODELS.values())))
```

```python
# ---- CONFIG: pick the two scripts to compare --------------------------
# Defaults if the dropdowns can't render in this viewer.
FILE_A = '02_Volume_Fib.pine'
FILE_B = '04_Pulse_mean_acc_strategy.pine'

HAS_WIDGETS = False
try:
    import ipywidgets as widgets
    dd_a = widgets.Dropdown(options=all_names, value=FILE_A, description='A:',
                            layout=widgets.Layout(width='420px'),
                            style={'description_width': '30px'})
    dd_b = widgets.Dropdown(options=all_names, value=FILE_B, description='B:',
                            layout=widgets.Layout(width='420px'),
                            style={'description_width': '30px'})

    def _sync(*_):
        global FILE_A, FILE_B
        FILE_A, FILE_B = dd_a.value, dd_b.value

    dd_a.observe(_sync, 'value')
    dd_b.observe(_sync, 'value')
    HAS_WIDGETS = True
    display(widgets.VBox([
        dd_a, dd_b,
        widgets.HTML('<i>verdict / neighbours / contribution update automatically — '
                     're-run the chunk cells manually after changing the selection</i>'),
    ]))
except Exception as e:
    print(f'ipywidgets unavailable ({e}) — edit FILE_A/FILE_B above instead')
    print('\n'.join(all_names))
```

```python
def cosine(a, b):
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-12))

def verdict(sim):
    if sim >= ei.REMIX_THRESHOLD:
        return 'REMIX — reward split recommended'
    if sim >= ei.RELATED_THRESHOLD:
        return 'related — same design family'
    return 'distinct'

def show_comparison(a, b):
    print(f'A: {a}')
    print(f'B: {b}\n')
    print(f"{'model':<6} {'cosine':>8}   verdict")
    print('-' * 60)
    for label, vecs in MODELS.items():
        if a not in vecs or b not in vecs:
            missing = a if a not in vecs else b
            print(f'{label:<6}   n/a     ({missing} not embedded)')
            continue
        s = cosine(vecs[a], vecs[b])
        print(f'{label:<6} {s:>8.4f}   {verdict(s)}')

if HAS_WIDGETS:
    display(widgets.interactive_output(show_comparison, {'a': dd_a, 'b': dd_b}))
else:
    show_comparison(FILE_A, FILE_B)
```

## Nearest neighbours

For context: each script's top-8 most similar scripts in each model.
If A and B are true parent/child, they should be each other's #1 (or near-top)
in both models, and their neighbour lists should overlap heavily.

```python
def top_neighbours(vecs, name, k=8):
    sims = [(n, cosine(vecs[name], v)) for n, v in vecs.items() if n != name]
    return sorted(sims, key=lambda t: -t[1])[:k]

def show_neighbours(a, b):
    for label, vecs in MODELS.items():
        if a not in vecs or b not in vecs:
            continue
        print(f'\n=== {label} ===')
        na, nb = top_neighbours(vecs, a), top_neighbours(vecs, b)
        rank_a_in_b = next((i + 1 for i, (n, _) in enumerate(top_neighbours(vecs, b, k=len(vecs))) if n == a), None)
        rank_b_in_a = next((i + 1 for i, (n, _) in enumerate(top_neighbours(vecs, a, k=len(vecs))) if n == b), None)
        print(f'  B ranks #{rank_b_in_a} among A\'s neighbours | A ranks #{rank_a_in_b} among B\'s')
        print(f"  {'A neighbours':<45} {'B neighbours':<45}")
        for i in range(max(len(na), len(nb))):
            la = f'{na[i][0]} ({na[i][1]:.4f})' if i < len(na) else ''
            lb = f'{nb[i][0]} ({nb[i][1]:.4f})' if i < len(nb) else ''
            mark = ' <== OTHER' if (i < len(na) and na[i][0] == b) or (i < len(nb) and nb[i][0] == a) else ''
            print(f'  {la:<45} {lb:<45}{mark}')

if HAS_WIDGETS:
    display(widgets.interactive_output(show_neighbours, {'a': dd_a, 'b': dd_b}))
else:
    show_neighbours(FILE_A, FILE_B)
```

## Chunk-level comparison (live embedding)

The stored vectors are mean-pooled over overlapping chunks, which hides *where*
the two scripts overlap. This cell re-embeds both files chunk by chunk (comments
stripped, like `build`) and shows the pairwise chunk similarity. High-similarity
cells show which regions were copied or remixed.

Chunk vectors are centered with the same document-level `reference_mean`. This
is an approximation, and the whole-file thresholds were not calibrated for
chunks, so read the chunk verdicts as a rough guide.

**Cost:** one Ollama call per file. Fast on `0.6b`, slow on `4b`.
`LIVE_LABEL` must be a loaded model (a key in `MODELS`).

```python
LIVE_LABEL = '0.6b'  # '4b' is ~5-6x slower
LIVE_MODEL = MODEL_NAMES[LIVE_LABEL]

# Uses FILE_A / FILE_B synced from the dropdowns — re-run this cell
# (and the two below) after changing the selection.

ei.MODEL = LIVE_MODEL

src_a = (ei.INDICATOR_DIR / FILE_A).read_text(encoding='utf-8', errors='replace')
src_b = (ei.INDICATOR_DIR / FILE_B).read_text(encoding='utf-8', errors='replace')

chunks_a = ei.chunk_source(ei.strip_comments(src_a))
chunks_b = ei.chunk_source(ei.strip_comments(src_b))
print(f'{FILE_A}: {len(chunks_a)} chunks | {FILE_B}: {len(chunks_b)} chunks')

mu = MEANS[LIVE_LABEL]
vecs_a = np.array([ei.center(v, mu) for v in ei.embed_texts(chunks_a)])
vecs_b = np.array([ei.center(v, mu) for v in ei.embed_texts(chunks_b)])
sim_matrix = vecs_a @ vecs_b.T  # (len a) x (len b), centered cosine
print(f'chunk-pair similarity: max={sim_matrix.max():.4f}  mean={sim_matrix.mean():.4f}')
```

```python
try:
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(max(6, len(chunks_b)), max(4, len(chunks_a) * 0.5)))
    im = ax.imshow(sim_matrix, cmap='viridis', vmin=-0.2, vmax=1.0, aspect='auto')
    ax.set_xlabel(f'{FILE_B} chunk #')
    ax.set_ylabel(f'{FILE_A} chunk #')
    ax.set_title(f'Chunk-pair centered cosine — {LIVE_MODEL}')
    for i in range(sim_matrix.shape[0]):
        for j in range(sim_matrix.shape[1]):
            ax.text(j, i, f'{sim_matrix[i, j]:.2f}', ha='center', va='center',
                    fontsize=8, color='white' if sim_matrix[i, j] < 0.6 else 'black')
    fig.colorbar(im, ax=ax, label='centered cosine')
    plt.tight_layout()
    plt.show()
except ImportError:
    print('matplotlib not installed — text heatmap:')
    hdr = '      ' + ''.join(f'{j:>7}' for j in range(sim_matrix.shape[1]))
    print(hdr)
    for i in range(sim_matrix.shape[0]):
        print(f'{i:>4}  ' + ''.join(f'{sim_matrix[i, j]:>7.3f}' for j in range(sim_matrix.shape[1])))
```

```python
# Region-level summary in BOTH directions.
# A->B: which parts of A reappear in B. B->A: which parts of B are new (e.g. added strategy logic).
# Uniformly high -> full-file remix. Patchy high -> partial reuse.
def region_table(sim, src_name, dst_name):
    print(f'\n{src_name} -> {dst_name}')
    print(f'{"chunk":>7} {"best match":>10} {"sim":>7}  region verdict')
    for i, (s, j) in enumerate(zip(sim.max(axis=1), sim.argmax(axis=1))):
        v = 'copied' if s >= ei.REMIX_THRESHOLD else ('borrowed' if s >= ei.RELATED_THRESHOLD else 'original')
        print(f'{i:>7} {j:>10} {s:>7.3f}  {v}')

region_table(sim_matrix, FILE_A, FILE_B)
region_table(sim_matrix.T, FILE_B, FILE_A)
```

## Per-dimension contribution (where the score comes from)

A raw bar chart of `a[i]` vs `b[i]` is noise: dimensions are unlabeled and unordered. The vectors here are **centered** and unit-normalized, so `cos(a,b) = Σ a[i]·b[i]`, and the element-wise product shows which latent features carry the similarity *beyond* the shared Pine Script component.

On raw (uncentered) vectors this chart was dominated by the corpus-mean dimensions, so it showed "both are Pine code" rather than lineage. Centering removes that.

- positive bars: features both scripts have more than the average script
- negative bars: features that pull the pair apart
- `(a[i]−b[i])²` shows per-dimension disagreement (`||a−b||² = 2 − 2cos`)

For reference, the cell also prints A's similarity to its nearest *other* script, so you can see how far the pair stands above the next-best match.

```python
CONTRIB_LABEL = '0.6b'  # must be a key in MODELS

def show_contrib(a, b):
    vecs = MODELS[CONTRIB_LABEL]
    va, vb = vecs[a], vecs[b]
    contrib = va * vb                      # sums to centered cos(a, b) — vectors are unit-norm
    diff2 = (va - vb) ** 2                 # per-dim disagreement; sums to 2 - 2cos
    order = np.argsort(-np.abs(contrib))   # most influential dims first
    pos, neg = contrib[contrib > 0].sum(), contrib[contrib < 0].sum()

    # baseline: A vs its most similar script that is neither A nor B
    others = [(n, float(va @ v)) for n, v in vecs.items() if n not in (a, b)]
    base_name, base_sim = max(others, key=lambda t: t[1])

    print(f'{CONTRIB_LABEL}: centered cos={contrib.sum():.4f}   2-2cos={diff2.sum():.4f}')
    print(f'  agreeing mass={pos:+.4f}   disagreeing mass={neg:+.4f}')
    print(f'  top-20 dims carry {contrib[order[:20]].sum() / contrib.sum():.1%} of the score')
    print(f'  baseline: A vs nearest other script {base_name} = {base_sim:.4f}')

    try:
        import matplotlib.pyplot as plt
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
        ax1.bar(np.arange(len(contrib)), contrib[order], width=1.0)
        ax1.axhline(0, color='k', lw=0.5)
        ax1.set_title(f'{a} vs {b} — per-dim contribution a[i]*b[i] '
                      f'(centered, sorted by |contrib|, {CONTRIB_LABEL})')
        ax1.set_ylabel('contribution to cosine')
        ax2.bar(np.arange(len(diff2)), diff2[order], width=1.0, color='tomato')
        ax2.set_title('per-dim disagreement (a[i]-b[i])² (same dim order)')
        ax2.set_ylabel('squared diff')
        ax2.set_xlabel('dimension rank (not dimension id)')
        plt.tight_layout()
        plt.show()
    except ImportError:
        print('matplotlib not installed — printing top-20 dims instead:')
        for d in order[:20]:
            print(f'  dim {d:>4}: a={va[d]:+.4f} b={vb[d]:+.4f} '
                  f'contrib={contrib[d]:+.5f} diff2={diff2[d]:.5f}')

if HAS_WIDGETS:
    display(widgets.interactive_output(show_contrib, {'a': dd_a, 'b': dd_b}))
else:
    show_contrib(FILE_A, FILE_B)
```
